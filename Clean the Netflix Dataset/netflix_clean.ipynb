```python
import pandas as pd
data=pd.read_csv("cleaned_netflix.csv")
print(data.info())
print(data.describe())
print(data.head())

print(data.isnull().sum())

data["min"]=data["min"].fillna("Unknown")


```

```python
data["Seasons"]=data["Seasons"].fillna("Unknown")

```

```python
data["cast"]=data["director"].fillna("Unknown")
```

```python
data["director"]=data["director"].fillna("Unknown")
```

```python
data["min"]=data["duration"].str.extract(r"(\d+)\s*min")
```

```python
data["Seasons"]=data["duration"].str.extract(r"(\d+)\s*Seasons")
```

```python
print(data)
```

```python
data["release_year"]=pd.to_datetime(data["release_year"])
```

```python
data["date_added"]=pd.to_datetime(data["date_added"],errors="coerce")
```

```python
data.to_csv("netflix_cleaned.csv")
print(data)
```
